```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../Data/Financial_Fraud_Data.csv')
```

```python
df.info()
df.describe()
```

```python
df.head()
```

```python
fraudVsflagged = df.groupby('isFraud')['isFlaggedFraud'].sum()
print(fraudVsflagged)
fraudVsflagged.plot(kind='bar')
plt.xlabel('Actual Fraud (0 = No, 1 = Yes)')
plt.ylabel('Is Flagged')
plt.show()
```

```python
pd.crosstab(df['isFraud'], df['isFlaggedFraud'])
```

```python
fraudVstype = df.groupby('type')['isFraud'].count()
print(fraudVstype)
fraudVstype.plot(kind='bar')
plt.ylabel('Is Fraud')
plt.show()
```
