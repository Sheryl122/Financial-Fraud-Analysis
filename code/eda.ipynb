```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('../Data/Financial_Fraud_Data.csv')
```

```python
df.info()
df.describe()
```

```python
df.head()
```

### Univariate Analyis

```python
df['type'].value_counts().plot(kind='bar')
plt.xlabel('Transaction Type')
plt.ylabel('Count')
plt.title('Count by Transactions Type')
plt.show()
```

```python
df['amount'].hist(bins=50, range=(0, df['amount'].quantile(0.95)))
plt.xlabel('Amount')
plt.ylabel('Count')
plt.title('Transaction Amounts')
plt.show()
```

### Bivariate Analysis

```python
flaggedvsfraud = pd.crosstab(df['isFraud'], df['isFlaggedFraud'])
flaggedvsfraud
```

```python
sns.heatmap(flaggedvsfraud, annot=True, fmt='d', cmap='Reds', cbar=False)
```

```python
fraudVstype = df.groupby('type')['isFraud'].sum()
print(fraudVstype)
fraudVstype.plot(kind='bar')
plt.title('Fraud Count by Type')
plt.ylabel('Is Fraud')
plt.show()
```

```python
fraudVstypepercent = df.groupby('type')['isFraud'].mean().round(5) * 100
fraudVstypepercent
```

```python
avg_amount = df.groupby('isFraud')['amount'].mean()
print(avg_amount)

avg_amount.plot(kind='bar')
plt.show()
```

On average the transaction amounts that are actually Fraudulent are significantly larger in value.

### Multivariate

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
```

Transaction type seems to be a big indicator of fraud, however none of these fields show a strong correlation to fraud in isolation. Fraud detection will likely require the combination of multiple field in consideration.
