## Which insights did you gain from your EDA?

Based on my exploratory analysis, fraud is heavily found the transaction types, TRANSFER and CASH_OUT, while the other types showed little to almost no fraud. I also noticed that fraudulent transactions had a higher amount of money transfered/cashed out on average despite weak correlation discovered on the heatmap. I also found that the isFlaggedFraud only caught 16 out 8197 fraudulent cases which is less than 1%, meaning virtually almost 100% of the cases were missed by the the isFlaggedFraud rule.
